# Part 1: Producing the Data (10%)
In this task, we will implement an Apache Kafka producer in Python to simulate real-time data streaming. Spark and parallel data processing should not be used in this section, as we are simulating roadside edge sensors that lack distributed processing capabilities. However, you can use the standard Python library to preprocess the data.  

1. Every 5 seconds, load a batch of records from `streaming_traffic_data.csv` and `streaming_parking_bays.csv`. You should keep a pointer in the files reading process and advance it per read. The data must be read in chronological order.  
   **Notes:**  
   - If you performed some strategies to reduce the amount of data by aggregation or filtering, you can do it here using Python standard library or pandas (no Spark).  
   - A batch refers to a minimum set of data that your model requires to make a prediction, for example, one `time_bin` for a site.  
2. Add the current timestamp (`traffic_ts`, `parking_ts`) to the records in the batch to represent the real-time event transmission time.  
3. Send your batch of traffic sensor data to two Kafka topics with appropriate names (e.g., `traffic_stream`, `parking_stream`).
