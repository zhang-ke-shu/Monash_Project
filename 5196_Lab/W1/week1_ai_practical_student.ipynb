## **The Challenge: Build a Small Scraping Pipeline**

The target is [Books to Scrape](https://books.toscrape.com/), a sandbox created for web-scraping practice.

Start from:

`https://books.toscrape.com/catalogue/page-1.html`

### **Required Result**

1. Visit the **first two catalogue pages** by following the site's **next-page link**. Do not hard-code the second page URL.
2. Collect all 20 product cards from each catalogue page.
3. Follow each product link to its detail page.
4. Build one tidy row per book with these columns:

| Column | Required value and type |
|---|---|
| `title` | full detail-page title as `str` |
| `price_gbp` | price without `£`, as `float` |
| `rating` | convert One-Five to integer `1-5` |
| `availability_count` | number extracted from text such as `In stock (22 available)`, as `int` |
| `upc` | product UPC as `str` |
| `category` | final category link in the breadcrumb before the product title, as `str` |
| `product_url` | absolute URL as `str` |

5. Store the rows in a Pandas DataFrame.
6. Validate the result in code:
   - exactly **40 rows**;
   - `upc` and `product_url` are unique;
   - required columns contain no missing values;
   - price is floating point, while rating and availability are integers;
   - `price_gbp > 0`, rating is between 1 and 5, and availability is at least 0.
7. Create a category summary with `book_count`, `average_price_gbp`, and `average_rating`.
8. Save:
   - `output/books.csv`
   - `output/category_summary.csv`
9. Display the first five book rows and the category summary.

### **Technical Constraints**

- Use `requests`, `beautifulsoup4`, `pandas`, `re`, `time`, `pathlib`, and `urllib.parse.urljoin`.
- Use a `requests.Session`, a descriptive `User-Agent`, a timeout, and `raise_for_status()`.
- Add a short delay between requests.
- Split the pipeline into small functions for fetching, parsing catalogue pages, parsing detail pages, validating, and saving.
- Do not use Selenium or Playwright; this practice site is static HTML.
