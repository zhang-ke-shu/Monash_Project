## Question 1 (7.5 marks)

Let $X\,{\sim}\,\mathcal{IG}\left(\theta: \left(\mu, \lambda\right)\right),\;\,\forall\, \mu>0 \text{ and } \lambda>0$. This means
the random varible $X$ follows the **inverse Gaussian distribution** with the set $\left(\theta: \left(\mu, \lambda\right)\right)$ acting as the parameters of said distribution. Given that we observe a sample of size $n$ that is independently and identically distributed from this distribution ([i.i.d](https://en.wikipedia.org/wiki/Independent_and_identically_distributed_random_variables)), $\mathbf{x}=$ $\left(x_{1}, \ldots, x_{n}\right)$, please find the [maximum likelihood estimate](https://en.wikipedia.org/wiki/Maximum_likelihood_estimation) for $\mu$ and $\lambda$, that is $\mu_{\text{MLE}}$ and $\lambda_{\text{MLE}}$. The probability density function (**PDF**) is as follows:

$$
f(x \mid \mu, \lambda)=\left\{\begin{array}{cc}
\left(\frac{\lambda}{2 \pi x^{3}}\right)^{1 / 2} e^{\frac{-\lambda(x-\mu)^{2}}{2 \mu^{2} x}}, & x>0 \\
0, & x \leq 0
\end{array}\right.
$$

### ANSWER

### Step 1 Likehood:
$$
L(\mu,\lambda)
=
\prod_{i=1}^n
\left[
\left(\frac{\lambda}{2\pi x_i^3}\right)^{1/2}
\exp\left(
-\frac{\lambda (x_i-\mu)^2}{2\mu^2 x_i}
\right)
\right]
$$


### Step 2 Negative log-likelihood:
$$
\begin{aligned}
-\log L(\mu,\lambda)
&=
-\log \prod_{i=1}^n f(x_i\mid\mu,\lambda) \\
&=
-\sum_{i=1}^n \log f(x_i\mid\mu,\lambda) \\
&=
-\sum_{i=1}^n
\left[
\frac{1}{2}\log\left(\frac{\lambda}{2\pi x_i^3}\right)
-
\frac{\lambda (x_i-\mu)^2}{2\mu^2 x_i}
\right] \\
&=
-\frac{n}{2}\log\lambda
+\frac{n}{2}\log(2\pi)
+\frac{3}{2}\sum_{i=1}^n \log x_i
+\frac{\lambda}{2\mu^2}\sum_{i=1}^n \frac{(x_i-\mu)^2}{x_i}
\end{aligned}
$$


### Step 3 Find the MLE for $\mu$ and $\lambda$:
### $\mu_{\text{MLE}}$:
$$
\begin{aligned}
-\log L(\mu,\lambda)
&=
-\frac{n}{2}\log\lambda
+\frac{n}{2}\log(2\pi)
+\frac{3}{2}\sum_{i=1}^n \log x_i
+\frac{\lambda}{2\mu^2}\sum_{i=1}^n \frac{(x_i-\mu)^2}{x_i}
\end{aligned}
$$

$$
=
-\frac{n}{2}\log\lambda
+\frac{n}{2}\log(2\pi)
+\frac{3}{2}\sum_{i=1}^n \log x_i
+\frac{\lambda}{2\mu^2}
\left(
\sum_{i=1}^n x_i - 2n\mu + \mu^2\sum_{i=1}^n \frac{1}{x_i}
\right)
$$

$$
=
-\frac{n}{2}\log\lambda
+\frac{n}{2}\log(2\pi)
+\frac{3}{2}\sum_{i=1}^n \log x_i
+\frac{\lambda}{2}
\left(
\frac{1}{\mu^2}\sum_{i=1}^n x_i
-\frac{2n}{\mu}
+\sum_{i=1}^n \frac{1}{x_i}
\right)
$$

$$
\frac{\partial}{\partial \mu}\bigl[-\log L(\mu,\lambda)\bigr]
=
\frac{\lambda}{2}
\left(
-\frac{2}{\mu^3}\sum_{i=1}^n x_i
+\frac{2n}{\mu^2}
\right)
$$

$$
=
\lambda
\left(
-\frac{1}{\mu^3}\sum_{i=1}^n x_i
+\frac{n}{\mu^2}
\right)
$$

$$
=
\frac{\lambda}{\mu^3}
\left(
-\sum_{i=1}^n x_i + n\mu
\right)
$$

$$
\frac{\partial}{\partial \mu}\bigl[-\log L(\mu,\lambda)\bigr]=0
$$

$$
\frac{\lambda}{\mu^3}
\left(
-\sum_{i=1}^n x_i + n\mu
\right)=0
$$

$$
-\sum_{i=1}^n x_i + n\mu = 0
$$

$$
n\mu = \sum_{i=1}^n x_i
$$

$$
\hat{\mu}_{\mathrm{MLE}}
=
\frac{1}{n}\sum_{i=1}^n x_i
=
\bar{x}
$$

### $\lambda_{\text{MLE}}$:
$$
\frac{\partial}{\partial \lambda}\bigl[-\log L(\mu,\lambda)\bigr]
=
-\frac{n}{2\lambda}
+
\frac{1}{2\mu^2}\sum_{i=1}^n \frac{(x_i-\mu)^2}{x_i}
$$

$$
\frac{\partial}{\partial \lambda}\bigl[-\log L(\mu,\lambda)\bigr]=0
$$

$$
-\frac{n}{2\lambda}
+
\frac{1}{2\mu^2}\sum_{i=1}^n \frac{(x_i-\mu)^2}{x_i}
=0
$$

$$
-\frac{n}{\lambda}
+
\frac{1}{\mu^2}\sum_{i=1}^n \frac{(x_i-\mu)^2}{x_i}
=0
$$

$$
\frac{1}{\mu^2}\sum_{i=1}^n \frac{(x_i-\mu)^2}{x_i}
=
\frac{n}{\lambda}
$$

$$
\hat{\lambda}_{\mathrm{MLE}}
=
\frac{n\mu^2}{\sum_{i=1}^n \frac{(x_i-\mu)^2}{x_i}}
$$

$$
\hat{\lambda}_{\mathrm{MLE}}
=
\frac{n\bar{x}^2}{\sum_{i=1}^n \frac{(x_i-\bar{x})^2}{x_i}}
$$

$$
\therefore\quad
\hat{\mu}_{\mathrm{MLE}}=\bar{x},
\qquad
\hat{\lambda}_{\mathrm{MLE}}
=
\frac{n\bar{x}^2}{\sum_{i=1}^n \frac{(x_i-\bar{x})^2}{x_i}}.
$$

## Question 2 (7.5 marks)
Suppose that we know that the random variable $X{\,\sim\,}\mathcal{Dist}(\mu=\theta, \sigma^2=\theta^2)$ follows the PDF given below:

$$
\begin{equation}
f\left(x|\theta\right)=
\begin{cases}
\frac{1}{\theta} \exp(-\frac{x}{\theta}) & x > 0 \\
0 & \text{otherwise.}
\end{cases}
\end{equation}
$$

Given a sample of $n$ [i.i.d](https://en.wikipedia.org/wiki/Independent_and_identically_distributed_random_variables) observations $\mathbf{x}=$ $\left(x_{1}, \ldots, x_{n}\right)$ from this distribution, please answer the following questions:

**(a)** Derive the MLE estimator for $\theta$, i.e., $\hat{\theta}_{\text{MLE}}$, and show that it is unbiased. [2.5 Marks]

**(b)** Find an estimator with better Mean-squared error (MSE) (i.e smaller MSE) compared to the $\hat{\theta}_{\text{MLE}}$ obtained from (a). [5 Marks]

### ANSWER


### (a) Derive the MLE estimator for $\theta$ and show that it is unbiased


$$
L(\theta)
=
\prod_{i=1}^n f(x_i\mid \theta)
=
\prod_{i=1}^n
\left[
\frac{1}{\theta}\exp\left(-\frac{x_i}{\theta}\right)
\right].
$$


$$
L(\theta)
=
\theta^{-n}\exp\left(-\frac{1}{\theta}\sum_{i=1}^n x_i\right).
$$

$$
\log L(\theta)
=
\log\left[
\theta^{-n}\exp\left(-\frac{1}{\theta}\sum_{i=1}^n x_i\right)
\right]
$$

$$
=
-n\log\theta
-\frac{1}{\theta}\sum_{i=1}^n x_i.
$$


$$
\frac{d}{d\theta}\log L(\theta)
=
-\frac{n}{\theta}
+
\frac{1}{\theta^2}\sum_{i=1}^n x_i.
$$

Set the derivative equal to zero:

$$
-\frac{n}{\theta}
+
\frac{1}{\theta^2}\sum_{i=1}^n x_i
=
0.
$$

Multiply both sides by $\theta^2$:

$$
-n\theta+\sum_{i=1}^n x_i=0.
$$

Thus,

$$
n\theta=\sum_{i=1}^n x_i,
$$

so the MLE is

$$
\hat{\theta}_{\mathrm{MLE}}
=
\frac{1}{n}\sum_{i=1}^n x_i
=
\bar{x}.
$$

Therefore,

$$
\hat{\theta}_{\mathrm{MLE}}=\bar{x}
$$

Now show that it is unbiased.

Given that

$$
E[X]=\theta,
\qquad
\mathrm{Var}(X)=\theta^2,
$$

we have

$$
E[\hat{\theta}_{\mathrm{MLE}}]
=
E[\bar{X}]
=
E\left[\frac{1}{n}\sum_{i=1}^n X_i\right]
$$

$$
=
\frac{1}{n}\sum_{i=1}^n E[X_i]
=
\frac{1}{n}\sum_{i=1}^n \theta
=
\theta.
$$

Hence,

$$
E[\hat{\theta}_{\mathrm{MLE}}]=\theta,
$$

so $\hat{\theta}_{\mathrm{MLE}}$ is unbiased.

$$
\hat{\theta}_{\mathrm{MLE}}=\bar{X}\text{ is unbiased.}
$$

### (b) Find an estimator with smaller MSE than $\hat{\theta}_{\mathrm{MLE}}$

Since

$$
E[X]=\theta,
\qquad
\mathrm{Var}(X)=\theta^2,
$$

so 

$$
E[\bar{X}]=\theta,
\qquad
\mathrm{Var}(\bar{X})=\frac{\theta^2}{n}.
$$

Because $\hat{\theta}_{\mathrm{MLE}}=\bar{X}$ is unbiased,

$$
\mathrm{MSE}(\hat{\theta}_{\mathrm{MLE}})
=
\mathrm{Var}(\bar{X})
=
\frac{\theta^2}{n}.
$$

Consider the estimator

$$
\tilde{\theta}=c\bar{X},
\qquad c\in\mathbb{R}.
$$

Then

$$
E[\tilde{\theta}]
=
E[c\bar{X}]
=
cE[\bar{X}]
=
c\theta,
$$

so its bias is

$$
\mathrm{Bias}(\tilde{\theta})
=
E[\tilde{\theta}]-\theta
=
(c-1)\theta.
$$

Also,

$$
\mathrm{Var}(\tilde{\theta})
=
\mathrm{Var}(c\bar{X})
=
c^2\mathrm{Var}(\bar{X})
=
c^2\frac{\theta^2}{n}.
$$

Therefore,

$$
\mathrm{MSE}(\tilde{\theta})
=
\mathrm{Var}(\tilde{\theta})
+
\bigl(\mathrm{Bias}(\tilde{\theta})\bigr)^2
$$

$$
=
c^2\frac{\theta^2}{n}
+
(c-1)^2\theta^2
$$

$$
=
\theta^2
\left(
\frac{c^2}{n}+(c-1)^2
\right).
$$

Let

$$
g(c)=\frac{c^2}{n}+(c-1)^2.
$$


$$
g'(c)=\frac{2c}{n}+2(c-1).
$$

Set $g'(c)=0$:

$$
\frac{2c}{n}+2(c-1)=0
$$

$$
\frac{c}{n}+c-1=0
$$

$$
c\left(1+\frac{1}{n}\right)=1
$$

$$
c=\frac{1}{1+\frac{1}{n}}
=\frac{n}{n+1}.
$$

Hence, choose

$$
\tilde{\theta}
=
\frac{n}{n+1}\bar{X}.
$$

Now compute its MSE.

Its bias is

$$
\mathrm{Bias}(\tilde{\theta})
=
\left(\frac{n}{n+1}-1\right)\theta
=
-\frac{\theta}{n+1},
$$

and its variance is

$$
\mathrm{Var}(\tilde{\theta})
=
\left(\frac{n}{n+1}\right)^2\frac{\theta^2}{n}
=
\frac{n\theta^2}{(n+1)^2}.
$$

Thus,

$$
\mathrm{MSE}(\tilde{\theta})
=
\frac{n\theta^2}{(n+1)^2}
+
\frac{\theta^2}{(n+1)^2}
$$

$$
=
\frac{(n+1)\theta^2}{(n+1)^2}
=
\frac{\theta^2}{n+1}.
$$

Compare this with the MSE of the MLE:

$$
\mathrm{MSE}(\hat{\theta}_{\mathrm{MLE}})
=
\frac{\theta^2}{n},
\qquad
\mathrm{MSE}(\tilde{\theta})
=
\frac{\theta^2}{n+1}.
$$

Since

$$
\frac{\theta^2}{n+1}
<
\frac{\theta^2}{n},
$$


$$
\tilde{\theta}
=
\frac{n}{n+1}\bar{X}
$$

has smaller MSE than $\hat{\theta}_{\mathrm{MLE}}=\bar{X}$.

Therefore, an estimator with better MSE is

$$
\tilde{\theta}
=
\frac{n}{n+1}\bar{X}.
$$

## Question 3 (7.5 marks)
Suppose that we know that a random variable $X$ follows the distribution given below:

$$
f\left(x|\theta\right)= \frac{{2 \choose x}\theta^x\left(1-\theta\right)^{2-x}}{1-\left(1-\theta\right)^2}, \; x= \{1,2\}
$$

Imagine that we observe a sample of $\textbf{n}$ [i.i.d](https://en.wikipedia.org/wiki/Independent_and_identically_distributed_random_variables) random varaibles $\mathbf{x}=$ $\left(x_{1}, \ldots, x_{n}\right)$ and want to model them using this distribution. Please use the concept of maximum likelihood to estimate the parameter $\theta$.

### ANSWER

### Step 1 Likehood:

Let

$$
m=\sum_{i=1}^{n} I(x_i=2),
$$

where m is the number of $x_i$ equal to 2. Then the number of $x_i$ equal to 1 is

$$
n-m.
$$

The given probability mass function is

$$
f(x\mid\theta)
=
\frac{\binom{2}{x}\theta^x(1-\theta)^{2-x}}
{1-(1-\theta)^2},
\qquad x\in\{1,2\}.
$$

Simplify the denominator:

$$
1-(1-\theta)^2
=
1-(1-2\theta+\theta^2)
=
2\theta-\theta^2
=
\theta(2-\theta).
$$

For \(x=1\),

$$
f(1\mid\theta)
=
\frac{\binom{2}{1}\theta(1-\theta)}
{\theta(2-\theta)}
=
\frac{2\theta(1-\theta)}
{\theta(2-\theta)}
=
\frac{2(1-\theta)}{2-\theta}.
$$

For \(x=2\),

$$
f(2\mid\theta)
=
\frac{\binom{2}{2}\theta^2}
{\theta(2-\theta)}
=
\frac{\theta^2}
{\theta(2-\theta)}
=
\frac{\theta}{2-\theta}.
$$

The likelihood function is

$$
L(\theta)
=
\prod_{i=1}^{n} f(x_i\mid\theta).
$$

Since there are n-m $x_i$ equal to 1 and m $x_i$ equal to 2,

$$
L(\theta)
=
\left[
\frac{2(1-\theta)}{2-\theta}
\right]^{n-m}
\left[
\frac{\theta}{2-\theta}
\right]^m.
$$

$$
L(\theta)
=
\frac{
2^{n-m}(1-\theta)^{n-m}\theta^m
}
{(2-\theta)^n}.
$$

### Step 2 Negative log-likelihood:

$$
-\log L(\theta)
=
-\log
\left[
\frac{
2^{n-m}(1-\theta)^{n-m}\theta^m
}
{(2-\theta)^n}
\right].
$$


$$
-\log L(\theta)
=
-(n-m)\log 2
-
(n-m)\log(1-\theta)
-
m\log\theta
+
n\log(2-\theta).
$$

### Step 3 Find the MLE for $\theta$:

Differentiate the negative log-likelihood with respect to $\theta$:

$$
\frac{d}{d\theta}[-\log L(\theta)]
=
\frac{n-m}{1-\theta}
-
\frac{m}{\theta}
-
\frac{n}{2-\theta}.
$$

Set the derivative equal to zero:

$$
\frac{n-m}{1-\theta}
-
\frac{m}{\theta}
-
\frac{n}{2-\theta}
=
0.
$$

Multiply both sides by$\theta(1-\theta)(2-\theta)$

$$
(n-m)\theta(2-\theta)
-
m(1-\theta)(2-\theta)
-
n\theta(1-\theta)
=
0.
$$


$$
2(n-m)\theta-(n-m)\theta^2
-
m(2-3\theta+\theta^2)
-
n\theta+n\theta^2
=
0.
$$

$$
2(n-m)\theta-(n-m)\theta^2
-
2m+3m\theta-m\theta^2
-
n\theta+n\theta^2
=
0.
$$


$$
(n+m)\theta-2m=0.
$$

$$
(n+m)\theta=2m.
$$

$$
\hat{\theta}_{\mathrm{MLE}}
=
\frac{2m}{n+m}.
$$


Since $x_i\in\{1,2\}$, m is the number of $x_i$ equal to 2

$$
\sum_{i=1}^{n}x_i
=
n+m.
$$

$$
m
=
\sum_{i=1}^{n}x_i-n.
$$

Therefore,

$$
\hat{\theta}_{\mathrm{MLE}}
=
\frac{2m}{n+m}
=
\frac{
2\left(\sum_{i=1}^{n}x_i-n\right)
}
{
\sum_{i=1}^{n}x_i
}.
$$

Since

$$
\sum_{i=1}^{n}x_i=n\bar{x}.
$$

Thus,

$$
\hat{\theta}_{\mathrm{MLE}}
=
\frac{2(n\bar{x}-n)}{n\bar{x}}
=
\frac{2(\bar{x}-1)}{\bar{x}}.
$$

Final result:

$$
\hat{\theta}_{\mathrm{MLE}}
=
\frac{2m}{n+m}
=
\frac{
2\left(\sum_{i=1}^{n}x_i-n\right)
}
{
\sum_{i=1}^{n}x_i
}
=
\frac{2(\bar{x}-1)}{\bar{x}}.
$$

## Question 4 (7.5 marks)
Suppose that we know that the random variable $X$ follows the PDF given below:

$$
\begin{equation}
    f\left(x|\theta\right)= 
    \begin{cases}
         e^{-(x-\theta)} & x\geq\theta \\
         0 & \text{otherwise.}
    \end{cases}
\end{equation}
$$

Given a sample of $n$ [i.i.d](https://en.wikipedia.org/wiki/Independent_and_identically_distributed_random_variables) observations $\mathbf{x}=$ $\left(x_{1}, \ldots, x_{n}\right)$ from this distribution, please answer the following questions:

**(a)** Derive the MLE estimator for $\theta$, i.e., $\hat{\theta}_{\text{MLE}}$. [4.5 Marks]

**(b)** Show that the estimator $\hat{\theta} = \overline{X} - 1$ (where $\overline{X} = \frac{1}{n}\sum_{i=1}^n X_i$) is an unbiased and consistent estimator for the given distribution. [3 Marks]

### ANSWER

### (a) Derive the MLE estimator for $\theta$, i.e., $\hat{\theta}_{\text{MLE}}$
### Step 1 Likelihood:
For
$$
f(x\mid\theta)
=
\begin{cases}
e^{-(x-\theta)}, & x\geq \theta,\\
0, & \text{otherwise}.
\end{cases}
$$

all $x_i$ must satisfy

$$
x_i\geq \theta,
\qquad i=1,\dots,n.
$$

So

$$
\theta \leq \min(x_1,\dots,x_n).
$$

The likelihood function is

$$
L(\theta)
=
\prod_{i=1}^{n} f(x_i\mid\theta).
$$


$$
L(\theta)
=
\begin{cases}
\prod_{i=1}^{n} e^{-(x_i-\theta)}, & \theta\leq \min(x_1,\dots,x_n),\\
0, & \theta> \min(x_1,\dots,x_n).
\end{cases}
$$

For $\theta\leq \min(x_1,\dots,x_n)\$,

$$
L(\theta)
=
\prod_{i=1}^{n} e^{-(x_i-\theta)}
=
e^{-\sum_{i=1}^{n}(x_i-\theta)}.
$$

$$
L(\theta)
=
e^{-\left(\sum_{i=1}^{n}x_i-n\theta\right)}
=
e^{-\sum_{i=1}^{n}x_i+n\theta}.
$$

### Step 2 Log-likelihood:

For $\theta\leq \min(x_1,\dots,x_n)\$,

$$
\log L(\theta)
=
-\sum_{i=1}^{n}x_i+n\theta.
$$

For $\theta> \min(x_1,\dots,x_n)\$,

$$
L(\theta)=0.
$$

### Step 3 Find the MLE for $\theta\$:


$$
\log L(\theta)
=
-\sum_{i=1}^{n}x_i+n\theta,
$$


$$
\frac{d}{d\theta}\log L(\theta)=n, n>0, \theta\leq \min(x_1,\dots,x_n).
$$

Thus the log-likelihood is increasing with $\theta$,

Since the $\theta\leq \min(x_1,\dots,x_n)$, maximum $\theta$ is $\min(x_1,\dots,x_n)$, so

$$
\hat{\theta}_{\mathrm{MLE}}
=
\min(x_1,\dots,x_n).
$$

# Part 2 Confidence Interval Estimation & Central Limit Theorem (30 marks) No Programming Involved
**WARNING:** If it is not explicitly stated, please assume the 95% confidence or 5% significant level.

## Question 1 (5 marks)
The [SETU](https://www.monash.edu/ups/setu) score of FIT units is known to follow a $\mathcal{N}(\mu=4$, $\sigma^2=0.25)$ distribution. You take a sample of the units and check their last semester's SETU. How many units do you have to sample to have a 95% confidence interval for $\mu$ with width 0.1?

### ANSWER

For two-sided tail
$$
X \sim N(\mu=4,\sigma^2=0.25)
$$


$$
\sigma=\sqrt{0.25}=0.5
$$

$$
z_{0.975}=1.96
$$

The confidence interval for $\mu$ is

$$
\bar{X}\pm z_{0.975}\frac{\sigma}{\sqrt{n}}
$$

The width of the confidence interval is

$$
2z_{0.975}\frac{\sigma}{\sqrt{n}}
$$

Set the width equal to 0.1:

$$
2z_{0.975}\frac{\sigma}{\sqrt{n}}=0.1
$$


$$
2(1.96)\frac{0.5}{\sqrt{n}}=0.1
$$

$$
\frac{1.96}{\sqrt{n}}=0.1
$$


$$
\sqrt{n}=\frac{1.96}{0.1}=19.6
$$

$$
n=(19.6)^2=384.16
$$
Since n must be an integer, round up,

$$
n=385
$$


## Question 2 (5 marks)
You do a poll to see what fraction $p$ of the students participated in the FIT5197 SETU survey. You then take the average frequency of all surveyed people as an estimate $\hat p$  for $p$. Now it is necessary to ensure that there is at least $99\%$ certainty that the difference between the surveyed rate $\hat p$   and the actual rate $p$ is not more than $5\%$. At least how many people should take the survey?

### ANSWER

For two-sided tail

$$

### Additional methodology discussion

The current methodology has a design issue. The report says it compares members against non-members, but the actual method compares the same customers before and after becoming members.

This before-after comparison may be affected by other factors, such as seasonality, inflation, promotions, changes in customer income, or general changes in shopping behaviour over time. Therefore, the increase in spending cannot be fully attributed to the membership subscription itself.

A better methodology is to use a control group.

One possible design is:

- Randomly select a group of customers who became members.
- Select a comparable group of customers who did not become members.
- Compare their spending over the same time period.
- Use similar customers in both groups, based on previous spending level, shopping frequency, and demographic information.

A stronger method is a difference-in-differences design:

$$
(\text{member after} - \text{member before})
-
(\text{non-member after} - \text{non-member before})
$$

This method compares how spending changed for members relative to how spending changed for non-members. It helps control for general time effects and gives a more reliable estimate of the effect of membership.

````r
samplingHelper <- function(n) {
  # Check whether n is a valid input
  if (!is.numeric(n) || length(n) != 1 || is.na(n)) {
    stop("n must be a single numeric value.")
  }
  
  if (n <= 0) {
    stop("n must be positive.")
  }
  
  if (n != floor(n)) {
    stop("n must be an integer.")
  }
  
  # Generate n uniform random numbers from 0 to 1
  p <- runif(n, min = 0, max = 1)
  
  # Create an empty vector to store generated random numbers
  numbers <- numeric(n)
  
  # First part of the quantile function: 0 <= p < 1/8
  index_left <- p < 1/8
  
  numbers[index_left] <- (-1 + sqrt(8 * p[index_left])) / 4
  
  # Second part of the quantile function: 1/8 <= p <= 1
  index_right <- p >= 1/8
  
  numbers[index_right] <- (7/4) * (1 - sqrt((8/7) * (1 - p[index_right])))
  
  return(numbers)
}
````

### (c)

````r
# Generate 99,999 random numbers from the target distribution
set.seed(123)
samples <- samplingHelper(99999)

# Plot histogram with 100 bins
hist(
  samples,
  breaks = 100,
  probability = TRUE,
  main = "Histogram of Generated Samples with Theoretical Density",
  xlab = "x",
  ylab = "Density",
  col = "lightgray",
  border = "white"
)

# Define the theoretical density function
theoretical_density <- function(x) {
  ifelse(
    x >= -1/4 & x < 0,
    4 * x + 1,
    ifelse(
      x >= 0 & x < 7/4,
      -4/7 * x + 1,
      0
    )
  )
}

# Create x values for plotting the theoretical curve
x_values <- seq(-1/4, 7/4, length.out = 1000)

# Overlay the theoretical density curve
lines(
  x_values,
  theoretical_density(x_values),
  col = "red",
  lwd = 2
)
````

### (d)
The sampling method used here is inverse transform sampling. First, I derived the cumulative distribution function \(F(x)\) from the given density function \(f(x)\). Then, I solved \(F(x)=p\) to obtain the quantile function \(Q(p)=F^{-1}(p)\). Since the density is piecewise-defined, the quantile function is also piecewise-defined, with a cutoff at \(p=1/8\).

To generate random numbers from \(f(x)\), I first generate random values

$$
p\sim U(0,1).
$$

Then I transform each value using the quantile function:

$$
X=Q(p).
$$

If $0\le p<1/8\$, I use

$$
Q(p)=\frac{-1+\sqrt{8p}}{4}.
$$

If $1/8\le p\le 1\$, I use

$$
Q(p)=\frac74\left(1-\sqrt{\frac87(1-p)}\right).
$$

This method works because of the probability integral transform. If X has CDF, then $F(X)\sim U(0,1)$. Conversely, if $p\sim U(0,1)$, then $Q(p)=F^{-1}(p)$ has CDF. Therefore, applying the quantile function to uniform random numbers produces samples following the required density $f(x)$.

### (e)
The PDF is

$$
f(x)=e^{-\pi x^2}, -\infty<x<\infty.
$$

This density corresponds to a normal distribution with

$$
\mu=0,
\sigma^2=\frac{1}{2\pi}.
$$

Thus,

$$
X\sim N\left(0,\frac{1}{2\pi}\right).
$$

The cumulative distribution function is

$$
F(x)=P(X\le x).
$$

Using the standard normal CDF \(\Phi\),

$$
F(x)
=
\Phi\left(
\frac{x-0}{\sqrt{1/(2\pi)}}
\right).
$$

Since

$$
\sqrt{\frac{1}{2\pi}}
=
\frac{1}{\sqrt{2\pi}},
$$

we have

$$
F(x)
=
\Phi(\sqrt{2\pi}x).
$$

Therefore,

$$
F(x)=\Phi(\sqrt{2\pi}x).
$$

The quantile function is

$$
Q(p)=F^{-1}(p).
$$

Let

$$
p=F(x).
$$

Then

$$
p=\Phi(\sqrt{2\pi}x).
$$

Apply \(\Phi^{-1}\) to both sides:

$$
\Phi^{-1}(p)=\sqrt{2\pi}x.
$$

Thus,

$$
x=\frac{\Phi^{-1}(p)}{\sqrt{2\pi}}.
$$

Therefore,

$$
Q(p)=\frac{\Phi^{-1}(p)}{\sqrt{2\pi}},
\qquad 0<p<1.
$$

### (f)

````r
samplingHelper2 <- function(n) {
  # Check whether n is a valid input
  if (!is.numeric(n) || length(n) != 1 || is.na(n)) {
    stop("n must be a single numeric value.")
  }
  
  if (n <= 0) {
    stop("n must be positive.")
  }
  
  if (n != floor(n)) {
    stop("n must be an integer.")
  }
  
  # Generate n uniform random numbers from 0 to 1
  p <- runif(n, min = 0, max = 1)
  
  # Apply the quantile function
  numbers <- qnorm(p) / sqrt(2 * pi)
  
  return(numbers)
}
````

### (g)

````r
set.seed(123)

samples_new <- samplingHelper2(99999)

hist(
  samples_new,
  breaks = 100,
  probability = TRUE,
  main = "Histogram of Generated Samples with Theoretical Density",
  xlab = "x",
  ylab = "Density",
  col = "lightgray",
  border = "white"
)

theoretical_density <- function(x) {
  exp(-pi * x^2)
}

x_values <- seq(min(samples_new), max(samples_new), length.out = 1000)

lines(
  x_values,
  theoretical_density(x_values),
  col = "red",
  lwd = 2
)
````

The histogram closely follows the theoretical curve $f(x)=e^{-\pi x^2}$, showing that the generated samples match the target distribution well. This density has the same shape as a Gaussian distribution. In fact,

$$
f(x)=e^{-\pi x^2}
$$

corresponds to

$$
N\left(0,\frac{1}{2\pi}\right).
$$

Therefore, the distribution is a Gaussian distribution with mean 0 and variance $\frac{1}{2\pi}$.
